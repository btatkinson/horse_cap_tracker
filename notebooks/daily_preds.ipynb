```python

def split_helper(x):
    if len(x) > 0:
        if ',' in x:
            x = x.split(',')
        if type(x) != list:
            x = x.replace('.0','')
            x = str(int(x))
    return x

def convert_multiple(path):
    df = pd.read_csv(path) ### CHECK HEADER
#     print(df)
    df['Date'] = pd.to_datetime(df['Date'].copy())
#     print(df['Date'])
#     df = df.loc[df['Date']==date]
#     print(df.head())
#     df.columns=['Capper','Race','Best','Top','A','B']
    
    pick_dict = {}
    for index, row in df.iterrows():
        capper = row['Capper']
        race = row['Race']
        if capper not in pick_dict:
            pick_dict[capper] = {}
        if type(race)== int:
            pass
        elif 'R' in race:
            race = race.replace('R','')
            race = int(race)
        elif 'Race ' in race:
            race = race.replace('Race ','')
            race = int(race)
        if not pd.isnull(row['Best']):
            best = str(row['Best'])
        else:
            best = 'nan'
        if not pd.isnull(row['Top']):
            top = str(row['Top'])
        else:
            top = 'nan'
        if not pd.isnull(row['A']):
            A = str(row['A'])
        else:
            A = 'nan'
        if not pd.isnull(row['B']):
            B = str(row['B'])
        else:
            B = 'nan'
        best = best.strip()
        top = top.strip()
        A = A.strip()
        B = B.strip()
        if 'nan' in best:
            best = []
        if 'nan' in top:
            top = []
        if 'nan' in A:
            A = []
        if 'nan' in B:
            B = []
        
        best = split_helper(best)
        top = split_helper(top)
        A = split_helper(A)
        B = split_helper(B)
        if type(A)==list:
            for a in A:
                if a in top:
                    A.remove(a)
                if a in best:
                    A.remove(a)
        elif type(A)==str:
            if type(top)==list:
                if A in top:
                    A = []
            elif A == top:
                A = []
            if type(best)==list:
                if A in best:
                    A = []
            elif A == best:
                A = []
                
        if type(best) == list:
            for b in best:
                if b in top:
                    top.remove(b)
        pick_dict[capper][race] = {
            'Best':[best] if type(best)!=list else best,
            'Top':[top] if type(top)!=list else top,
            'A':[A] if type(A)!=list else A,
            'B':[B] if type(B)!=list else B
        }
    return pick_dict
path = 'C://Users/Blake/Downloads/Racing 02-17-2024 - fairgrounds.csv'
multi_pick_dict = convert_multiple(path)
print(multi_pick_dict)

fbg = multi_pick_dict['Fat Bald Guy']
# vanessa = multi_pick_dict['Vanessa']
# phillip = multi_pick_dict['Phillip']
# ptf = multi_pick_dict['PTF']
# jk = multi_pick_dict['JK']
# boar_matt = multi_pick_dict['Boar_Matt']
kelsch = multi_pick_dict['Kelsch']
david_aragona = multi_pick_dict['David Aragona']
# mike_beer = multi_pick_dict['Mike Beer']
# andy_serling = multi_pick_dict['Andy Serling']
# anthony_stabile = multi_pick_dict['Anthony Stabile']
# illman = multi_pick_dict['Dan Illman']
# # derosa = multi_pick_dict['Derosa']

holmes = multi_pick_dict['Holmes']
# jason_o = multi_pick_dict['Jason O']

blake = multi_pick_dict['Blake']
goaty = multi_pick_dict['Goaty']
burly = multi_pick_dict['Burly']
craig_milkowski = multi_pick_dict['Milkowski']
# # marty = multi_pick_dict['Marty']
# duke = multi_pick_dict['Duke']
# somich = multi_pick_dict['Somich']
# n_tamm = multi_pick_dict['Nick Tammaro']
# benny = multi_pick_dict['Benny']
# chris = multi_pick_dict['Chris']
# cnote = multi_pick_dict['CNote']
# kid = multi_pick_dict['Kid']
# rd_jared = multi_pick_dict['Racing Dudes Jared']
# rd_holfren = multi_pick_dict['Racing Dudes Holfren']

```

```python

def convert_to_float(frac_str):
    try:
        return float(frac_str)
    except ValueError:
        num, denom = frac_str.split('/')
        try:
            leading, num = num.split(' ')
            whole = float(leading)
        except ValueError:
            whole = 0
        frac = float(num) / float(denom)
        return whole - frac if whole < 0 else whole + frac

def load_upc(track):
    entries = pd.read_csv(f'../data/upc_entries/{track}.csv')
    meta = pd.read_csv(f'../data/upc_meta/{track}.csv')
    entries['running'] = np.where(entries['horse_name'].isin([
        "Practical Move","Algiers (IRE)","Exaulted","A Mo Reay","Pretty Mischievous","Bolshoi Ballet (IRE)","Get Smokin",
        "Arcangelo","Bradsell (GB)","One Timer","Lane Way","Twilight Gleaming (IRE)","American Theorem",
        
    ]), 0, entries['running'].copy())
    
    entries['ml_odds'] = entries['ml_odds'].apply(convert_to_float)
    entries = entries.loc[entries['running']==1].reset_index(drop=True)
    entries['implied_odds'] = entries['ml_odds'].copy().apply(lambda x: 1/(1+x))
    entries['implied_sum'] = entries.groupby(['date','track','race'])['implied_odds'].transform('sum')
    entries['implied_odds'] = entries['implied_odds'].copy()/entries['implied_sum'].copy()
    entries = entries.drop(columns=['implied_sum'])
    entries['race'] = entries['race'].astype(int)
#     for col in ['win_paid','place_paid','show_paid']:
#         entries[col] = entries[col].replace('[\$,]', '', regex=True).astype(float)
    entries['horse'] = entries['horse'].astype(str).str.strip()
    return entries, meta



entries, meta = load_upc(track)

```
